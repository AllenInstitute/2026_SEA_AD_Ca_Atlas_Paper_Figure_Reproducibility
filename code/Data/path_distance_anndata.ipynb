```python
import pandas as pd
import spatialdata as sd
from src.utils import read_obs
from datetime import datetime
import scanpy as sc
import geopandas as gpd
```

```python
current_datetime = datetime.now()
formatted_date = current_datetime.strftime("%Y-%m-%d")
```

```python
adata = sc.read_h5ad("/root/capsule/data/combined_adata/CaH_Xenium.2026-01-07.h5ad")
```

```python
cell_polygons = gpd.read_parquet("/root/capsule/data/polygon_parquets_1/seaad_cah_cell_segmentation_polygons.2026-02-05.parquet")
```

```python
cell_polygons = cell_polygons.set_geometry("geometry")
```

# Abeta plaque distance

```python
plaque_polygons = gpd.read_parquet("/root/capsule/data/polygon_parquets_1/seaad_cah_plaque_polygons.2026-02-05.parquet")
```

```python
plaque_polygons = plaque_polygons.set_geometry("geometry")
```

```python
plaque_barcodes = plaque_polygons["barcode"].unique().tolist()
plaque_distance_gpds = []
for barcode in plaque_barcodes:
    print(f"Processing barcode: {barcode}")
    reference_polygons = plaque_polygons[plaque_polygons["barcode"] == barcode]
    query_polygons = cell_polygons[cell_polygons["barcode"] == barcode]
    idx, distance = reference_polygons.geometry.sindex.nearest(
        query_polygons.geometry, 
        return_all = False, 
        return_distance=True
        )
    query_polygons['nearest_plaque_distance'] = distance
    plaque_distance_gpds.append(query_polygons)
```

```python
cell_plaque_distance_gpd = pd.concat(plaque_distance_gpds)
```

```python
plaque_adata = adata[cell_plaque_distance_gpd.index.tolist(), :].copy()
plaque_adata.obs = plaque_adata.obs.merge(
    cell_plaque_distance_gpd[["nearest_plaque_distance"]] * 0.2125,
    left_index=True,
    right_index=True,
    how="left"
)
```

```python
plaque_adata.write_h5ad(f"/root/capsule/results/seaad_cah_plaque_distance_adata.{formatted_date}.h5ad")
```
