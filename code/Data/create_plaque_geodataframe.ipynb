```python
from src.utils import sdata_read_polygons
from anndata import read_h5ad
from pandas import read_csv, concat, DataFrame
from pathlib import Path
import spatialdata as sd
```

```python
neuropath_obs = read_csv("/root/capsule/data/Supplemental Tables/Supplemental Table 7.csv", index_col = 0)
```

```python
high_plaque_donors = neuropath_obs[neuropath_obs['percent 6e10 positive area'] > 0.1].index
```

```python
high_plaque_donors
```

```python
adata =  read_h5ad("/root/capsule/data/adata/CaH_Xenium_filtered.2026-07-23.h5ad")
```

```python
adata_high_path = adata[adata.obs['Donor ID'].isin(high_plaque_donors) & adata.obs['Used in analysis']]
```

```python
adata_high_path.obs.index = adata_high_path.obs['barcode'].astype(str) + "_" + adata_high_path.obs['cell_id'].astype(str)
```

```python
adata_high_path.write_h5ad("/results/CaH_Xenium_filtered_high_plaque.2026-07-23.h5ad")
```

```python
barcodes = adata_high_path.obs['barcode'].unique()
```

```python
cell_gdfs = {}
plaque_gdfs = {}
for barcode in barcodes:
    print(barcode)
    path = Path(f"/root/capsule/data/spatialdata/{barcode}_with_vs200_CAH_processed.zarr")
    sdata = sd.read_zarr(path)
    plaque_gdf = sd.transform(sdata['plaque_boundaries'], to_coordinate_system='global')
    plaque_gdfs[barcode] = plaque_gdf
    cell_gdf = sd.transform(sdata['cell_boundaries'], to_coordinate_system = 'global')
    cell_gdf.index = barcode + '_' + cell_gdf.index
    cell_gdfs[barcode] = cell_gdf
```

```python
total_cell_gdf = concat(cell_gdfs, )
total_cell_gdf.index = total_cell_gdf.index.droplevel(0)
total_cell_gdf.to_parquet("/results/SEAAD_CaH_Cell_Boundaries.20260812.parquet")
```

```python
total_cell_df = DataFrame({
    "x":total_cell_gdf.centroid.x,
    "y":total_cell_gdf.centroid.y,
    "area":total_cell_gdf.area
})
```

```python
total_cell_df.to_csv("/results/SEAAD_CaH_Cell_Location.20260812.csv")
```

```python
total_plaque_gdf = concat(plaque_gdfs, )
total_plaque_gdf.index = total_plaque_gdf.index.get_level_values(0) + "_" + total_plaque_gdf['label'].astype(str)
total_plaque_gdf = total_plaque_gdf.drop(columns = 'label')
total_plaque_gdf.to_parquet("/results/SEAAD_CaH_Plaque_Boundaries.20260812.parquet")
```

```python
total_plaque_df = DataFrame({
    "x":total_plaque_gdf.centroid.x,
    "y":total_plaque_gdf.centroid.y,
    "area":total_plaque_gdf.area
})
```

```python
total_plaque_df.to_csv("/results/SEAAD_CaH_Plaque_Location.20260812.csv")
```
