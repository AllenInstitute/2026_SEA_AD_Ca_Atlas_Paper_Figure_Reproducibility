```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from src.utils import latexify_xlabel, save_figure
```

```python
plt.rcParams["font.size"] = 18
plt.rcParams['figure.dpi'] = 600
plt.rcParams['figure.figsize'] = (10, 10)
```

```python
donor_df = pd.read_csv("/root/capsule/data/Supplemental Tables/Supplemental Table 7.csv", index_col = 0)
```

```python
from scipy.stats import spearmanr
for idx, i in enumerate(['CPS_AT8', 'CPS_6E10']):
    fig, ax = plt.subplots()
    for j, color in {'percent AT8 positive area':"#762a83", 'percent 6e10 positive area':"#1b7837"}.items():
        r, p = spearmanr(donor_df[i], donor_df[j])
        plt.scatter(donor_df[i], donor_df[j], label = j.split(" ")[1] + f" $r_s =$ {r:.2f}", color = color, )

    plt.legend()
    plt.ylabel("Pathology % Area")
    plt.xlabel(latexify_xlabel(i))
    save_figure(fig, f"/scratch/Figure_3_{i}.svg", dpi = 600, formats = ("svg", ))
    plt.show()
```
